```python
import requests
import base64
import json
```

## Configure API Request

Set up the API endpoint and request payload.

```python
# API endpoint
api_url = "https://6b9eca0d4657.ngrok-free.app/generate"

# Request payload
payload = {
    "text": "bedframe",
    "seed": 42,
    "num_inference_steps": 5,
    "guidance_scale": 5.0,
    "texture": False,
    "face_count": 40000,
    "type": "glb"
}

print(f"API URL: {api_url}")
print(f"Payload: {json.dumps(payload, indent=2)}")
```

## Send Request to API

Make a POST request to generate the 3D model.

```python
# Send POST request
print("Sending request to API...")
response = requests.post(api_url, json=payload)

# Check if request was successful
if response.status_code == 200:
    print("✓ Request successful!")
    response_data = response.json()
    print(f"Request ID: {response_data.get('request_id', 'N/A')}")
else:
    print(f"✗ Request failed with status code: {response.status_code}")
    print(f"Response: {response.text}")
    raise Exception(f"API request failed: {response.status_code}")
```

## Save Generated Files

Decode the base64-encoded data and save the image and 3D model files.

```python
# Save the image
image_filename = response_data['image']['filename']
with open(f'output/{image_filename}', 'wb') as f:
    f.write(base64.b64decode(response_data['image']['data']))
print(f"✓ Image saved: {image_filename}")

# Save the model
model_filename = response_data['model']['filename']
with open(f'output/{model_filename}', 'wb') as f:
    f.write(base64.b64decode(response_data['model']['data']))
print(f"✓ Model saved: output/{model_filename}")

print("\nGeneration complete! Files saved successfully.")
```

## Display the Generated Image

Show the generated image inline in the notebook.

```python
from IPython.display import Image, display

# Display the generated image
display(Image(filename=image_filename))
```

## Verify GLB File

Check if the generated GLB file is valid and can be read.

```python
import os

# Check file size and magic bytes
file_size = os.path.getsize(model_filename)
print(f"Model file: output/{model_filename}")
print(f"File size: output/{file_size:,} bytes")

# Read first 12 bytes to verify GLB header
with open(model_filename, 'rb') as f:
    header = f.read(12)
    
    # GLB files should start with "glTF" magic bytes (0x46546C67)
    magic = header[0:4]
    version = int.from_bytes(header[4:8], 'little')
    length = int.from_bytes(header[8:12], 'little')
    
    print(f"\nGLB Header:")
    print(f"  Magic: {magic} (should be b'glTF')")
    print(f"  Version: {version} (should be 2)")
    print(f"  Length: {length:,} bytes")
    
    if magic == b'glTF' and version == 2:
        print("\n✓ GLB file appears valid!")
    else:
        print("\n✗ GLB file may be corrupted!")
        print(f"  First 12 bytes (hex): {header.hex()}")
```
